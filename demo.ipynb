# Language Models
A small synthetic example of the maintained 2026 modules. Selected Fall 2022 student implementations are in `historical/`.

```python
from ngrams import train,generate
vocabulary,counts,probs=train(['the sun rises','the moon rises','the sun shines'])
for order in (2,3):
    print(order,generate(vocabulary=vocabulary,probs=probs,order=order,seed=0))
```
